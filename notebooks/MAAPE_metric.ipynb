```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import wilcoxon
import os

classes = ['all', 'Smooth', 'Intermittent', 'Erratic', 'Lumpy']
# Загружаем evaluation-файлы

for cls in classes:
    fname = f'forecasts_{cls}.csv'
    if os.path.exists(fname):
        df = pd.read_csv(fname)

        model_cols = df.columns[df.columns.get_loc('y') + 1:]

        evaluator = RegressionMetric()

        # все MAAPE для каждой комбинации (модель, unique_id)
        records = []
        for unique_id, group in df.groupby('unique_id'):
            y_true = group['y'].values
            for model in model_cols:
                y_pred = group[model].values
                maape = evaluator.mean_arctangent_absolute_percentage_error(y_true, y_pred)
                records.append({'Model': model, 'unique_id': unique_id, 'MAAPE': maape})

        # промежуточные результаты
        maape_df = pd.DataFrame(records)

        # средние и медианы по моделям
        agg_results = maape_df.groupby('Model')['MAAPE'].agg(['mean', 'median']).reset_index()
        agg_results = agg_results.sort_values('mean')


        agg_results = agg_results.rename(columns={'mean': 'MAAPE_mean', 'median': 'MAAPE_median'})

        print(f"MAAPE для класса {cls}")
        print(agg_results)
        agg_results.to_csv(f'maape_aggregated_{cls}.csv', index=False)
    else:
        print(f"Файл {fname} не найден!")

```
